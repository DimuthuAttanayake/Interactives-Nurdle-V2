```python
import pandas as pd

# load the full dataset
df = pd.read_csv("nurdle-patrol-export.csv")
```

```python
# keep only lat, lon, and amount collected columns
df_sliced = df.loc[:, ["Latitude", "Longitude", "Amount Collected"]]
df_sliced
```

```python
# filter to rows where amount collected is greater than 0
df_sliced = df_sliced[df_sliced["Amount Collected"] > 0]
df_sliced
```

```python
# save the sliced data as a JSON file
df_sliced.to_json("sliced_data.json", orient="records", indent=2)
```
